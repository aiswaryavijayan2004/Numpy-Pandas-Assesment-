Q16.Studene Performance Analysis

A .   Data Cleanng

```python

print("Missing values:")
print(df.isnull().sum())

num_cols = ['Age', 'Attendance', 'Python_Score', 'SQL_Score',
            'Pandas_Score', 'NumPy_Score', 'Final_Score']

df[num_cols] = df[num_cols].fillna(df[num_cols].mean())

# B. Feature Engineering
df['Average_Score'] = (
    df['Python_Score'] +
    df['SQL_Score'] +
    df['Pandas_Score'] +
    df['NumPy_Score']
) / 4

df['Pass_Status'] = np.where(
    df['Average_Score'] >= 50,
    'Pass',
    'Fail'
)

# C. Analysis

# 1. Top 5 students
print("Top 5 Students:")
print(df.sort_values('Average_Score', ascending=False).head(5))

# 2. Department having highest average score
department_avg = df.groupby('Department')['Average_Score'].mean()
print("\nDepartment with highest average score:")
print(department_avg.idxmax())

# 3. Average attendance by department
print("\nAverage Attendance by Department:")
print(df.groupby('Department')['Attendance'].mean())

# 4. Number of students passed and failed
print("\nPass and Fail Count:")
print(df['Pass_Status'].value_counts())

# D. Summary
summary = df.groupby('Department').agg(
    Total_Students=('Student_ID', 'count'),
    Average_Score=('Average_Score', 'mean'),
    Average_Attendance=('Attendance', 'mean'),
    Pass_Count=('Pass_Status', lambda x: (x == 'Pass').sum()),
    Fail_Count=('Pass_Status', lambda x: (x == 'Fail').sum())
)

print("\nSummary:")
print(summary)
```

#Q18. Advanced Employee Analysis

```python


import pandas as pd
import numpy as np

df = pd.read_csv('/content/drive/MyDrive/DSAB13/NumPy and Pandas Assesment/employee.csv')

df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')
df['Experience'] = pd.to_numeric(df['Experience'], errors='coerce')

#A.
salary = df['Salary'].dropna().values
perf = df['Performance_Score'].dropna().values

print("Mean Salary:", salary.mean())
print("Median Salary:", np.median(salary))
print("Std Salary:", salary.std())
print("Mean Performance:", perf.mean())

#B.
filtered = df[(df['Experience'] > 3) & (df['Performance_Score'] >= 80)]
print(filtered)
#C.
dept_summary = df.groupby('Department').agg(
    Average_Salary=('Salary', 'mean'),
    Average_Performance=('Performance_Score', 'mean'),
    Total_Projects=('Projects_Completed', 'sum'),
    Employee_Count=('Employee_ID', 'count')
)
print(dept_summary)
print("Highest Performance Dept:", dept_summary['Average_Performance'].idxmax())

#D.
flagged = df[(df['Salary'] > salary.mean()) & (df['Performance_Score'] < perf.mean())]
print(flagged[['Employee_ID', 'Department', 'Salary', 'Performance_Score']])

```
